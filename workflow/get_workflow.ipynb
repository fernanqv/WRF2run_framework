# Preprocessing Workflow Diagram (WPS + real)

This workflow diagram illustrates the preprocessing steps for the **Weather Research and Forecasting (WRF)** model using Python's `graphviz` library. The workflow is divided into four main stages: **WPS geogrid**, **WPS ungrib**, **WPS metgrid**, and **WRF real.exe**. Each stage includes key processing tasks, with relevant links to resources when applicable.

---

## Workflow Stages

### 1. **geogrid.exe and LCZ Processing**
- **Define domain** in `namelist.wps`.
- **Download static data**: [WRF WPS static data](https://www2.mmm.ucar.edu/wrf/users/download/get_sources_wps_geog.html)
- **Download LCZ information** from Google Earth: [LCZ dataset](https://developers.google.com/earth-engine/datasets/catalog/RUB_RUBCLIM_LCZ_global_lcz_map_latest)
- **Run `geogrid.exe`** to generate geogrid files.
- **Install `w2w` tool** from GitHub: [w2w repository](https://github.com/matthiasdemuzere/w2w)
- **Run `w2w`** for all `geo_em` files.

---

### 2. **ungrib.exe and GCM Processing**
- **If using ERA5**: download the reanalysis data.
- **If using GCM**: download NetCDF data and run `bctool`.
- **Link/create Vtable** in the WPS folder.
- **Run `ungrib.exe`** to extract meteorological fields.

---

### 3. **metgrid.exe and Lake Temperature**
- **Check SST interpolation** settings in `METGRID.TBL`.
- **Run `metgrid.exe`** to interpolate meteorological data onto the model grid.
- **Postprocess metgrid output** for lake temperature averaging (Python, 15-day running mean).

---

### 4. **real.exe and Aerosols**
- **Link `met_em` files** into the WRF/run directory.
- **Run `real.exe`** to initialize the WRF model.
- **Convert aerosol data** into WRF-compatible format.

---

### 5. **Start and End**
- Workflow begins at **START** and ends at **END**, with all intermediate steps represented as nodes in the directed graph.

---

The diagram is generated programmatically using **Graphviz** and rendered as a PNG and PDF file. Each node is color-coded:
- **Orange (#fdae61)**: Main step
- **Yellow (#fee08b)**: Sub-steps or tasks
- **Green (#b3e2cd)**: Start and End nodes
- **Blue links** indicate external references.

The rendered output provides a clear visual guide for the WRF preprocessing workflow.

```python
from graphviz import Digraph

# Create directed graph
dot = Digraph(comment='Preprocess Workflow', format='PDF')

# Global graph style
dot.attr(rankdir='TB', size='10,8')

# Start / End nodes
dot.node('start', label='<<b><font point-size="26"> START </font></b>>', shape='oval', style='filled', fillcolor='#b3e2cd', labelloc='c', align='center')

# -----------------------------
# WPS: Step 1: geogrid.exe
# -----------------------------
dot.node('wps1', label='<<b><font point-size="18">1. geogrid.exe and LCZ processing </font></b>>', shape='box', style='filled', fillcolor='#fdae61')
dot.node('wps1a', 'Define domain in namelist.wps', shape='box', style='filled', fillcolor='#fee08b')
dot.node('wps1b', 'Download static data', shape='box', style='filled', fillcolor='#fee08b',
         href='https://www2.mmm.ucar.edu/wrf/users/download/get_sources_wps_geog.html', fontcolor='blue')
dot.node('wps1c', 'Download LCZ info from Google Earth', shape='box', style='filled', fillcolor='#fee08b',
         href='https://developers.google.com/earth-engine/datasets/catalog/RUB_RUBCLIM_LCZ_global_lcz_map_latest', fontcolor='blue')
dot.node('wps1d', 'Run geogrid.exe', shape='box', style='filled', fillcolor='#fee08b')
dot.node('wps1e', 'Install w2w tool from a github repo', shape='box', style='filled', fillcolor='#fee08b',
        href='https://github.com/matthiasdemuzere/w2w', fontcolor='blue')
dot.node('wps1f', 'Run w2w for all geo_em files', shape='box', style='filled', fillcolor='#fee08b')
dot.edges([('start','wps1'), ('wps1','wps1a'), ('wps1a','wps1b'),
           ('wps1b','wps1c'), ('wps1c','wps1d'), ('wps1d','wps1e'), ('wps1e','wps1f')])

# -----------------------------
# WPS: Step 2: ungrib.exe
# -----------------------------
dot.node('wps2', label='<<b><font point-size="18">2. ungrib.exe and GCM processing </font></b>>', shape='box', style='filled', fillcolor='#fdae61')
dot.node('wps2a', 'If ERA5: download data', shape='box', style='filled', fillcolor='#fee08b')
dot.node('wps2b', 'If GCM: download NetCDF & run bctool', shape='box', style='filled', fillcolor='#fee08b')
dot.node('wps2c', 'Link/create Vtable in WPS folder', shape='box', style='filled', fillcolor='#fee08b')
dot.node('wps2d', 'Run ungrib.exe', shape='box', style='filled', fillcolor='#fee08b')

dot.edges([('wps1f','wps2'), ('wps2','wps2a'), ('wps2','wps2b'),
           ('wps2a','wps2c'), ('wps2b','wps2c'), ('wps2c','wps2d')])

# -----------------------------
# WPS: Step 3: metgrid.exe
# -----------------------------
dot.node('wps3', label='<<b><font point-size="18">3. metgrid.exe and lake temperature</font></b>>', shape='box', style='filled', fillcolor='#fdae61')
dot.node('wps3a', 'Check SST interpolation in METGRID.TBL', shape='box', style='filled', fillcolor='#fee08b')
dot.node('wps3b', 'Run metgrid.exe', shape='box', style='filled', fillcolor='#fee08b')
dot.node('wps3c', 'Postprocess metgrid (Python lake Tavg, 15-day mean)', shape='box', style='filled', fillcolor='#fee08b')

dot.edges([('wps2d','wps3'), ('wps3','wps3a'), ('wps3a','wps3b'), ('wps3b','wps3c')])

# -----------------------------
# WRF Real: Step 4: real.exe
# -----------------------------
dot.node('wrf1', label='<<b><font point-size="18">4. real.exe and aerosols</font></b>>', shape='box', style='filled', fillcolor='#fdae61')
dot.node('wrf1a', 'In WRF/run directory link met_em files', shape='box', style='filled', fillcolor='#fee08b')
dot.node('wrf1b', 'Run real.exe', shape='box', style='filled', fillcolor='#fee08b')
dot.node('wrf1c', 'Convert aerosol data into WRF format', shape='box', style='filled', fillcolor='#fee08b')

dot.edges([('wps3c','wrf1'), ('wrf1','wrf1a'), ('wrf1a','wrf1b'), ('wrf1b','wrf1c')])


# End node
dot.node('end', label='<<b><font point-size="26"> END </font></b>>', shape='oval', style='filled', fillcolor='#b3e2cd', labelloc='c', align='center')
dot.edge('wrf1c', 'end')

# Render
output_file = 'preprocess_workflow'
dot.render(output_file, view=True)
print(f"Preprocessing workflow diagram generated: {output_file}.png")
```

# WRF Run Workflow

This diagram describes the WRF **Run Workflow**, from creating AOD files, setting up run directories, to executing `wrf.exe`.

## Workflow Steps

### 1. Create AOD files
- **Tool to create AOD file on GitHub:** [aerosols4wrf](https://github.com/AEI-CORDyS/aerosols4wrf)

### 2. Set-up files in `WRF/run`
- Check soft-linked **emission file**  
- Check parameters in `URBPARM_LCZ.TBL` 
- Check `varlist.txt` for variables to be deleted  
- Update **dates** in `namelist.input`, `AOD_{domain}` files, CO2 in `MPTABLE`

### 3. Run WRF
- Execute `wrf.exe` in the run directory

---

## Notes
- Links to external tools are highlighted in blue in the diagram.  

---

Generated using **Graphviz** with HTML-like labels, bold fonts, and colored nodes for clarity.

```python
from graphviz import Digraph

# Create directed graph
dot = Digraph(comment='WRF Run Workflow', format='png')

# Global graph style
dot.attr(rankdir='TB', size='10,8')

# -----------------------------
# Start node
# -----------------------------
dot.node(
    'start',
    label='<<b><font point-size="26"> START </font></b>>',
    shape='oval',
    style='filled',
    fillcolor='#b3e2cd',
    labelloc='c',
    align='center'
)

# -----------------------------
# WRF Step 1: Create AOD files
# -----------------------------
dot.node(
    'wrf1',
    label='<<b><font point-size="18">1. Create AOD files</font></b>>',
    shape='box',
    style='filled',
    fillcolor='#fdae61'
)
dot.node(
    'wrf1a',
    'Tool to create AOD file on GitHub',
    shape='box',
    style='filled',
    fillcolor='#fee08b',
    href='https://github.com/AEI-CORDyS/aerosols4wrf',
    fontcolor='blue'
)
dot.edges([('start','wrf1'), ('wrf1','wrf1a')])

# -----------------------------
# WRF Step 2: Set-up files in WRF/run
# -----------------------------
dot.node(
    'wrf2',
    label='<<b><font point-size="18">2. Set-up files in WRF/run</font></b>>',
    shape='box',
    style='filled',
    fillcolor='#fdae61'
)
dot.node('wrf2a', 'Check soft-linked emission file', shape='box', style='filled', fillcolor='#fee08b')
dot.node('wrf2b', 'Check parameters in URBPARM_LCZ.TBL', shape='box', style='filled', fillcolor='#fee08b')
dot.node('wrf2c', 'Check varlist.txt for variables to be deleted', shape='box', style='filled', fillcolor='#fee08b')
dot.node('wrf2d', 'Update dates in namelist.input, AOD_{domain} files, CO2 in MPTABLE', shape='box', style='filled', fillcolor='#fee08b')
dot.edges([('wrf1a','wrf2'), ('wrf2','wrf2a'), ('wrf2a','wrf2b'), ('wrf2b','wrf2c'), ('wrf2c','wrf2d')])

# -----------------------------
# WRF Step 3: Run wrf.exe
# -----------------------------
dot.node(
    'wrf3',
    label='<<b><font point-size="18">3. Run wrf.exe</font></b>>',
    shape='box',
    style='filled',
    fillcolor='#fdae61'
)
dot.edges([('wrf2d','wrf3')])

# -----------------------------
# End node
# -----------------------------
dot.node(
    'end',
    label='<<b><font point-size="26"> END </font></b>>',
    shape='oval',
    style='filled',
    fillcolor='#b3e2cd',
    labelloc='c',
    align='center'
)
dot.edge('wrf3', 'end')

# -----------------------------
# Render
# -----------------------------
output_file = 'run_workflow'
dot.render(output_file, view=True)
print(f"Run workflow diagram generated: {output_file}.png")
```

# Postprocess Workflow Diagram

This diagram illustrates the **postprocessing workflow** for WRF outputs using the **pCMORizer** tool. It is generated programmatically using Python's `graphviz` library and highlights the main steps from repository access to final verification.

---

## Workflow Stages

### 1. **Repository Access & Setup**
- **Request GitLab access** (contact: K. Görgen).
- **Clone the josipa branch** from the [pCMORizer repository](https://icg4geo.icg.kfa-juelich.de/ExternalRepos/pCMORizer/).
- **Install dependencies**:
  - NetCDF
  - CDO
  - NCO
  - Python
  - Fortran compiler
- **Compile CMORizer** using `make`.

---

### 2. **Configure CMORizer**
- **Set paths** to `geo_em` files and output directories in `runctrl.current.nml_template*`.
- **Edit the script** `run_pCMORizer_in_quasi_parallel.sh` to specify input files.

---

### 3. **Run CMORizer**
- **Option 1**: Process a single variable.
- **Option 2**: Batch processing of multiple variables using a CSV list.
- **Option 3**: Daily aggregation using `pCMORizer_aggregate_per_day.sh`.

---

### 4. **Verification**
- **Check for missing values and NaNs** before deleting raw WRF output files to ensure data integrity.

---

### 5. **Start and End**
- The workflow begins at **START** and ends at **END**, with intermediate steps represented as nodes in the directed graph.

---

### Color Coding
- **Orange (#fdae61)**: Main workflow steps
- **Yellow (#fee08b)**: Sub-tasks
- **Green (#b3e2cd)**: Start and End nodes
- **Blue hyperlinks** indicate external resources.

---

The rendered diagram provides a clear visual guide to the **pCMORizer-based postprocessing workflow**, helping to track tasks from initial repository setup to final data verification.
